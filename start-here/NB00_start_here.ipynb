## How Colab works

A notebook is a sequence of cells. Text cells, like this one, explain. Code cells hold Python instructions: Click into a code cell and press Shift+Enter to run it. The output appears directly below. Run the cells in order from the top. If something goes wrong, the menu Runtime, then Restart session, starts again from a clean state. The first code cell below prints a line of text and the results of three calculations: `/` divides, and `**` raises to a power.

```python
# Python practice: a small example for learning the language
print("Hello from the start-here notebook")
print(2 + 3, 10 / 4, 2 ** 8)
```

The next cell defines a small helper, `check`, that compares an answer with the expected value and prints feedback, and the cell after it keeps the expected values in the table `ANSWERS`, with its code hidden. Run both once; the exercises below use them. Each exercise starts with `None`, the Python value for nothing, in the place of the answer, and `check` reports it as not answered yet. An exercise calls `check` without the expected value, so that the answer does not stand next to it, and `check` looks the value up in `ANSWERS`. Reading the code of `check` is optional, since the sections below introduce most of its pieces. Two pieces appear only here. `expected=None` and `tol=1e-6` give the two parameters default values, which apply when a call leaves them out. The block under `try` compares the answer as a number, and if the answer is not a number, the block under `except` compares it as it is.

```python
def check(name, value, expected=None, tol=1e-6):
    """Compares an answer with the expected value and prints immediate feedback. Without an expected value in the
    call, check takes the value and its tolerance from ANSWERS, the table of the next cell."""
    if expected is None:
        answer, tol = ANSWERS[name]
        expected = answer()
    if value is None:
        print(f"{name}: not answered yet. Replace None with your answer and run the cell again.")
        return False
    try:
        ok = abs(float(value) - float(expected)) <= tol * max(1.0, abs(float(expected)))
    except (TypeError, ValueError):
        ok = value == expected
    print(f"{name}: " + ("correct." if ok else "not yet. Compare your steps with the lecture and try again."))
    return ok
```

```python
#@title Expected answers of the exercises (run this cell; open it only after trying the exercises)
ANSWERS = {
    "Exercise 0.1": (lambda: 4.8, 1e-6),
    "Exercise 0.2": (lambda: 2, 1e-6),
    "Exercise 0.3": (lambda: 180.0, 1e-6),
    "Exercise 0.4": (lambda: 47.3, 1e-6),
}
```

## Variables and types

A variable is a name for a value, and `=` binds the value on its right to the name on its left. Numbers with a decimal point are floats, whole numbers are integers, text in quotation marks is a string, and `True` and `False` are booleans. `type` names the type of a value. The `f` before the quotation mark in the last line makes an f-string, which inserts the value of each expression in braces into the text: `:.0f` prints a number without decimals, and `:.2f` prints it with two.

```python
# Python practice: a small example for learning the language
voltage_V = 48.0
capacity_Ah = 60
chemistry = "lithium-ion"
charging = False
print(type(voltage_V), type(capacity_Ah), type(chemistry), type(charging))
print(f"A {chemistry} pack of {voltage_V:.0f} V and {capacity_Ah} Ah stores {voltage_V * capacity_Ah / 1000:.2f} kWh.")
```

**Exercise 0.1.** A pack of 48 V and 100 Ah stores how many kilowatt-hours? Store the number in `kwh`.

```python
kwh = None   # your answer
check("Exercise 0.1", kwh)
```

```python
#@title Reference solution (run this cell; open it only after trying)
kwh = 48 * 100 / 1000
check("Exercise 0.1", kwh)
```

## Lists, loops and decisions

A list holds values in order. A `for` loop runs its indented block once for every element, and an `if` statement runs its block only when the condition is true. `elif` checks a further condition when the ones before it were false, and `else` runs its block when none of them was true.

```python
# Python practice: a small example for learning the language
temperatures = [31.5, 38.2, 44.7, 47.3, 42.0]
for t in temperatures:
    if t > 46:
        print(t, "too hot, full cooling")
    elif t > 42:
        print(t, "warm, fan on")
    else:
        print(t, "normal")
```

**Exercise 0.2.** Count the readings in `temperatures` that are above 42 and store the count in `n_warm`.

```python
n_warm = None   # your answer
check("Exercise 0.2", n_warm)
```

```python
#@title Reference solution (run this cell; open it only after trying)
n_warm = sum(1 for t in temperatures if t > 42)
check("Exercise 0.2", n_warm)
```

The reference solution counts in one line. `sum(1 for t in temperatures if t > 42)` adds a 1 for every reading above 42: The part inside the round brackets is a loop written as an expression, called a generator expression.

## Functions

`def` defines a reusable piece of code with parameters and a `return` value. The text in triple quotation marks describes the function.

```python
# Python practice: a small example for learning the language
def heat_W(current_A, resistance_ohm):
    """Ohmic heat generated in the pack."""
    return current_A ** 2 * resistance_ohm

print(heat_W(40, 0.045), "W")
```

**Exercise 0.3.** Use `heat_W` to store the heat generated by 60 A in a pack of 0.05 ohm in `q60`.

```python
q60 = None   # your answer
check("Exercise 0.3", q60)
```

```python
#@title Reference solution (run this cell; open it only after trying)
q60 = heat_W(60, 0.05)
check("Exercise 0.3", q60)
```

## Three libraries: NumPy, pandas and Matplotlib

`import` loads a library. NumPy computes with whole arrays at once, pandas holds tables with named columns, and Matplotlib draws plots. `import numpy as np` loads NumPy under the short name `np`. `np.arange(0, 50, 10)` makes the array 0, 10, 20, 30, 40: It starts at 0, steps by 10 and stops before 50. The curly braces build a dictionary, which maps keys to values, and `pd.DataFrame` turns it into a table with one column per key. `table["minute"]` selects a column by its name. In the plot, `"o-"` draws dots joined by lines, `plt.axhline` draws a horizontal line, and a semicolon puts two statements on one line.

```python
# Python practice: a small example for learning the language
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

minutes = np.arange(0, 50, 10)
table = pd.DataFrame({"minute": minutes, "temperature": temperatures})
print(table)
plt.plot(table["minute"], table["temperature"], "o-")
plt.axhline(46, color="red", linestyle="--")
plt.xlabel("minute"); plt.ylabel("temperature (°C)"); plt.show()
```

**Exercise 0.4.** Use NumPy to store the largest temperature in `temperatures` in `t_max`.

```python
t_max = None   # your answer
check("Exercise 0.4", t_max)
```

```python
#@title Reference solution (run this cell; open it only after trying)
t_max = float(np.max(temperatures))
check("Exercise 0.4", t_max)
```
